Using a set of example pwa results [stored in csv's](../data/) to demonstrate the collection, processing, and plotting pipeline. We simply need to provide a parent directory where the data is stored, and our [pipeline cli](../../src/pwassist/cli.py) does the rest. We do specify one optional argument, the final state parity. Since our example is for $\omega(P=-1)~\pi (P=-1)$, our final state has a total parity of $P=+1$. This is not crucial, but allows for some of the coherent sum plot legends to be fully drawn.

```python
%%bash
parent_dir=$(dirname "$(realpath "$0")")
data_dir=$(builtin cd ${parent_dir}/../data && pwd)

pwassist ${data_dir} --final-state-parity 1
```
